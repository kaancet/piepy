# Aggregating session/hub data

The aggregator turns a trial table into a **tidy** per-condition estimate frame — one row
per `(group × metric)` with `value`, `ci_low`, `ci_high`, `n`. It is experiment-agnostic
(you name the grouping columns and the metrics) and the output is plotting-ready.

This notebook runs on **synthetic data** so it works anywhere; the final cell shows an
optional real-session example.

````python
import numpy as np
import polars as pl
import matplotlib.pyplot as plt

from piepy.core.registry import get_session_class
from piepy.stats import aggregate, Median, Mean, Rate, Count
````

````python
exp_name = "240627_KC148_detect_opto120_HVA__1P_KC"
sess = get_session_class("detection")(exp_name, load_flag=True)
df = sess.concatenate_runs("detection")
print(exp_name, "->", df.shape)
````

````python
df.head()
````

### Grouping by a single/multiple column and calculating the rate occurance of a feature

````python
agg = aggregate(df,
                group="stim_side",
                rate="opto_pattern",
                rate_of=0
                )
agg
````

````python
agg = aggregate(df,
                group=["stim_side","contrast"],
                rate="outcome",
                rate_of="hit"
                )
agg
````

### Using ```metric``` objects to calculate Rate

````python
agg = aggregate(df,
                group=["stim_side","contrast"],
                metrics=[Rate("outcome","hit")]
                )
agg
````

### Using multiple ```metric``` specs to aggregate multiple measurements

````python
agg = aggregate(df,
                group=["stim_type","opto_pattern","contrast","stim_side"],
                metrics=[Median("reaction_time"), Rate("outcome","hit"), Count()],
                sort=True)
agg
````

#### rate / value / stat / rate_of: shorthands for a single metric when ``metrics`` is None

````python
agg = aggregate(df,
                group=["stim_side","contrast"],
                value="reaction_time",
                stat="median",
                )
agg
````
